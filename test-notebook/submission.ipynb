```python
import os
import subprocess

print('=== GPU ===')
subprocess.run(['nvidia-smi', '--query-gpu=name,memory.total,memory.free', '--format=csv,noheader'])

print('\n=== /kaggle/input contents ===')
for root, dirs, files in os.walk('/kaggle/input'):
    level = root.replace('/kaggle/input', '').count(os.sep)
    indent = ' ' * 2 * level
    print(f'{indent}{os.path.basename(root)}/')
    if level < 3:
        subindent = ' ' * 2 * (level + 1)
        for file in files[:20]:
            fpath = os.path.join(root, file)
            size = os.path.getsize(fpath) if os.path.isfile(fpath) else 0
            print(f'{subindent}{file} ({size/1e6:.1f}MB)')
        if len(files) > 20:
            print(f'{subindent}... and {len(files)-20} more files')

print('\n=== Looking for model files ===')
for root, dirs, files in os.walk('/kaggle/input'):
    for f in files:
        if f == 'config.json':
            print(f'  CONFIG: {os.path.join(root, f)}')
        if f.endswith('.safetensors') or f.endswith('.bin'):
            fp = os.path.join(root, f)
            print(f'  MODEL: {fp} ({os.path.getsize(fp)/1e9:.2f}GB)')

print('\n=== Check /kaggle/model ===')
for p in ['/kaggle/model', '/kaggle/models']:
    if os.path.exists(p):
        print(f'{p} exists!')
        for f in sorted(os.listdir(p)):
            print(f'  {f}')
```
